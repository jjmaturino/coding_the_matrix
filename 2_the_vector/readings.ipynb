Vectors originated as a way to represent **directed quantities** in physical space — things that have both magnitude and direction.

A vector can be visualized as an **arrow**:
- **Length** = magnitude (how much)
- **Direction** = which way it points

`dict.setdefault(key, default)` returns the value for `key` if it exists, otherwise inserts `key` with `default` and returns `default`.

```python
d = {'a': 1}

d.setdefault('a', 99)  # returns 1 (key exists, no change)
d.setdefault('b', 99)  # returns 99 (key didn't exist, now d['b'] = 99)

```

It's handy for initializing missing keys:

```python
counts = {}
for word in ['apple', 'banana', 'apple']:
    counts.setdefault(word, 0)
    counts[word] += 1
# {'apple': 2, 'banana': 1}
```

Compared to `get()`:
- `get(key, default)` — returns default but **doesn't modify** the dict
- `setdefault(key, default)` — returns default **and inserts** it if missing

In equations, **symmetry** means you can swap the roles of variables/terms without changing the meaning or result.

**Example from the chapter:**

The line segment from \(u\) to \(v\) can be written as:

$$\{u + \alpha(v - u) : 0 \le \alpha \le 1\}$$

This is **asymmetric** — it treats \(u\) as the "starting point" and \(v\) differently.

The symmetric form is:

$$\{\alpha v + \beta u : \alpha + \beta = 1, \alpha, \beta \ge 0\}$$

Now \(u\) and \(v\) appear in the same way — you can swap them and the expression looks identical. Neither is privileged as "the starting point."

**Why symmetry matters:**

1. **Reveals structure** — symmetric forms often expose the true nature of the relationship
2. **Easier to generalize** — the symmetric convex combination formula extends naturally to 3+ points
3. **Less arbitrary** — no artificial choice of which point is "first"
4. **Often simpler proofs** — symmetry can halve the work (prove for one case, the other follows by symmetry)

In math what is the generlized term convex and affine, how are they related to one another, how do they differ?

##### 🤖Reply🤖<!-- SOLVEIT_SEPARATOR_7f3a9b2c -->

🧠🧠🧠🧠🧠🧠🧠🧠🧠

**Affine** and **convex** are related concepts that describe different types of geometric structures:

---

**Affine:**
- Preserves **lines** and **parallelism**
- No notion of "betweenness" or "inside"
- **Affine combination**: \(\sum \alpha_i v_i\) where \(\sum \alpha_i = 1\) (weights sum to 1, can be negative)
- **Affine set**: A set closed under affine combinations (i.e., contains all lines through any two of its points)
- Examples: lines, planes, hyperplanes

---

**Convex:**
- Preserves **betweenness** and has a notion of "inside"
- **Convex combination**: \(\sum \alpha_i v_i\) where \(\sum \alpha_i = 1\) **and** \(\alpha_i \ge 0\) (non-negative weights)
- **Convex set**: A set closed under convex combinations (i.e., contains all line segments between any two of its points)
- Examples: line segments, triangles, circles, cubes

---

**Relationship:**

$$\text{Convex} \subset \text{Affine}$$

Every convex combination is an affine combination (with the extra constraint that weights are non-negative).

| | Affine | Convex |
|---|--------|--------|
| Weights sum to 1 | ✓ | ✓ |
| Weights non-negative | ✗ | ✓ |
| Geometric result | Entire line/plane | Segment/interior |

**Intuition:** Affine lets you "extrapolate" beyond the points. Convex restricts you to "interpolate" between them.

**Parity** means whether a count is **even or odd**.

In \(GF(2)\):
- Parity **0** = even number of ones
- Parity **1** = odd number of ones

The dot-product of an all-ones vector with another vector gives you the **parity** of that vector:

Two ones → even → parity is **0**.

This is the basis for **parity check bits** used in error detection — you can detect if a single bit flipped because the parity changes.

````
def triangular_solve_n(rowlist, b):
    D = rowlist[0].D
    n = len(D)
    assert D == set(range(n))
    x = zero_vec(D)
    for i in reversed(range(n)): x[i] = (b[i] - rowlist[i] * x)/rowlist[i][i]
    
    return x
````

$$\sqrt{a} \cdot \sqrt{a} = a^{1/2} \cdot a^{1/2} = a^{1/2 + 1/2} = a^1 = a$$

This follows directly from the exponent addition rule: \(a^m \cdot a^n = a^{m+n}\).
