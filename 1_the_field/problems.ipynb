Algebra is the branch of mathematics that studies **structure** and **relationships** using symbols.

At its heart, algebra asks: "What rules does this system follow, and what can I deduce from those rules?"

There are different levels:

- **Elementary algebra** — what you learned in school: solving equations like 2x + 3 = 7, manipulating expressions with variables
- **Abstract algebra** — studying the *structures themselves*: groups, rings, fields (like GF(2)!)

The power of algebra is that once you prove something using the rules, it applies to *anything* that follows those rules — numbers, matrices, functions, symmetries, even rubik's cube moves!
