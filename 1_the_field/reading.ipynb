**Overloading** in Python works via *method dispatch*, not coercion. When you write `a + b`, Python calls `a.__add__(b)`. Each type defines its own `__add__` method, so the same `+` symbol invokes different code depending on the type.

plaintext ⊕ key = ciphertext
