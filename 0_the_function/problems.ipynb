Problem 0.8.1: increments(L)
input: list L of numbers
output: list of numbers in which the ith element is one plus the ith element of L.
Example: increments([1,5,7]) should return [2,6,8].

````

def increments(L: list[int]) -> list[int]: return list(map(lambda x: x + 1, L))
increments([1,5,7])
````

````
"""
Problem 0.8.2: cubes(L)
input: list L of numbers
output: list of numbers in which the ith element is the cube of the ith element of L.
Example: given [1, 2, 3] return [1, 8, 27].
"""
def cubes(L:list[int]) -> list[int]: return list(map(lambda x: x**3, L))
cubes([1, 2, 3])
````

##### 🤖Reply🤖<!-- SOLVEIT_SEPARATOR_7f3a9b2c -->

Great question! This is a key distinction in understanding invertibility.

For a function to be invertible, it must be **onto** (surjective), which means every element in the **codomain** must be mapped to by at least one element in the domain.

Here's the important part: the **image** (or range) of a function is the set of all actual outputs—what the function *actually* hits. The **codomain** is what we *declare* the outputs could be.

**For invertibility**: We need the image to **equal** the codomain. If the codomain contains elements that aren't in the image, the function is not onto, so it's not invertible.

**Example**: Consider f: ℝ → ℝ where f(x) = x². 
- The codomain is ℝ (all real numbers)
- The image is [0, ∞) (only non-negative reals)
- Since image ≠ codomain, it's not onto, so not invertible

But if we change it to f: ℝ → [0, ∞), now the codomain equals the image, making it onto (though we'd still need to fix the domain to make it one-to-one).

##### 🤖Reply🤖<!-- SOLVEIT_SEPARATOR_7f3a9b2c -->

🧠🧠🧠🧠🧠🧠🧠🧠🧠🧠🧠🧠

Exactly right! You can always make a function onto by restricting the codomain to be exactly the image. This is sometimes called the "canonical surjection" or just saying we're viewing the function as mapping to its range.

However, for **full invertibility**, you still need the function to be **one-to-one** (injective) as well. Restricting the codomain doesn't help with that part.

So your "artificially onto" trick solves half the invertibility problem. The other half—making sure no two inputs map to the same output—often requires restricting the **domain**.

Back to f(x) = x²: even if we set the codomain to [0, ∞), it's still not invertible because both x = 2 and x = -2 map to 4. We'd also need to restrict the domain to, say, [0, ∞) to get a proper inverse.
