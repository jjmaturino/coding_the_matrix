The term "binary" here means the operator takes **two operands** (two values to work with). For example, in `3 + 5`, the `+` operator is binary because it operates on both `3` and `5`.

Yes, exactly! The modulo operation connects to **congruence** in number theory.

Two integers *a* and *b* are said to be **congruent modulo n** (written as a ≡ b (mod n)) if they have the same remainder when divided by n. Equivalently, this means n divides (a - b).

Great questions! Let me tackle these one at a time:

**Are we showing the roundoff error?**
Yes! When you add 1 to 1e16 (which is 10,000,000,000,000,000), Python returns 1e16 unchanged. The 1 you added just "disappeared" because of limited precision.

**Why is it rounding off?**
Python uses **floating-point arithmetic** (specifically, the IEEE 754 double-precision standard). These floats store numbers using a fixed number of bits, so they can only represent about 15-17 significant decimal digits. When you have a huge number like 1e16, there aren't enough bits left to also track a tiny change like +1.

**How much can be stored?**
Python floats use 64 bits total:
- 1 bit for sign
- 11 bits for exponent (determines the range: roughly 10^-308 to 10^308)
- 52 bits for the mantissa (determines precision: about 15-17 decimal digits)

**Why was 'e' chosen for scientific notation?**
The 'e' stands for "exponent" and comes from early computing and calculator conventions. It's compact and unambiguous for representing powers of 10.

Task 0.5.30: Define a one-line procedure dict2list(dct,keylist) with this spec:
• input: dictionary dct, list keylist consisting of the keys of dct
• output: list L such that L[i] = dct[keylist[i]] for i = 0, 1, 2, . . . , len(keylist)− 1
• example: input dct={'a':'A', 'b':'B', 'c':'C'} and keylist=['b','c','a'],
output ['B', 'C', 'A']

````
def dict2list(dct, keylist) -> list[any]: return list(map(lambda i: dct[keylist[i]], range(len(keylist))))
````

Task 0.5.31: Define a one-line procedure list2dict(L, keylist) specified as follows:
• input: list L, list keylist of immutable items
• output: dictionary that maps keylist[i] to L[i] for i = 0, 1, 2, . . . , len(L)− 1
• example: input L=[’A’,’B’,’C’] and keylist=[’a’,’b’,’c’],
output {'a':'A', 'b':'B', 'c':'C'}
Hint: Use a comprehension that iterates over a zip or a range.

````
def list2dict(l, keylist): return {x[0]: x[1] for x in zip(keylist, l)}
````
