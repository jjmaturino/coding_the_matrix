**Lemma**: A proven statement that's typically used as a stepping stone to prove something more important (like a theorem). It's a "helper" result. Example: Lemma 0.3.20 about composition of invertible functions.

**Proof**: The logical argument that demonstrates why a statement (like a lemma or theorem) must be true, using definitions, axioms, and previously proven results.
