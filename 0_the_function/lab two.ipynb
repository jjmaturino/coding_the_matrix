````
def make_inverse_index(documents: list[str]) -> dict[str, set[int]]:
    pdcs = list(map(lambda d: re.findall(r'\w+', d), documents))

    res = {}
    for idx, d in enumerate(pdcs):
        for w in d:
            w = w.lower() if len(w) != 1 else w
            res.setdefault(w, set()).add(idx)

    return res
````

Good question about `reduce`!

**Syntax:**
```python
reduce(function, iterable, [initializer])
```

- `function`: takes 2 arguments (accumulator, next_item)
- `iterable`: the sequence to process
- `initializer`: optional starting value

**Why no initial value here?**

When you don't provide an initializer, `reduce` uses the first element of the iterable as the starting accumulator. So:

```python
reduce(lambda a, b: a | b, [{0, 1, 3}, {1, 2}])
```

1. First call: `a = {0, 1, 3}`, `b = {1, 2}` → returns `{0, 1, 2, 3}`
2. That becomes the final result (no more items)

You *could* provide `set()` as an initializer, but it's not necessary here since the first element is already a set.
