Great observation — they're actually two *different* metaphors doing different jobs:

- **"Under"** (as in "image *under* f") means "subjected to" or "acted upon by." The function f *acts on* x, so x passes *under* the operation of f. Think of it like going under a machine that transforms you.

- **"Over"** (as in "vector *over* a field") means "built on top of." The vector space is a structure that *sits on top of* the field as its foundation. The field provides the scalars, and everything else is constructed *above* it.

So the spatial metaphors are actually consistent: the field is the **ground floor**, and structures are built *over* it. Meanwhile, a function is a **machine**, and inputs pass *under* it to be transformed.

Both usages trace back to German mathematical tradition — "über" (over/above) was used by algebraists like Emmy Noether when describing algebraic structures built upon base fields/rings, and "unter" (under) appeared in analysis for function application. The conventions carried into English mathematical writing and stuck.
